```python
import pandas as pd

# Create a DataFrame
data = {
    "name": ["Ali", "John", "Sara"],
    "age": [20.0, 21.0, 19.0]
}

df = pd.DataFrame(data)

print("Before:")
print(df)
print(df.dtypes)

# Convert float to integer
df["age"] = df["age"].astype(int)

print("\nAfter:")
print(df)
print(df.dtypes)
```

```python
import numpy as np

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print(array1.dtype)
print(array2.dtype)

if array1.dtype == array2.dtype:
    print("The arrays have the same datatype")
else:
    print("The arrays have different datatypes")
```

```python
import numpy as np

array1 = np.array([1, 2, 3])
array2 = np.array([1.0, 2.0, 3.0])

print(array1.dtype)
print(array2.dtype)

if array1.dtype == array2.dtype:
    print("Same datatype")
else:
    print("Different datatypes")
```
